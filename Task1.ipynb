```python
# Task1: Text Normalization, Tokenization and Stopword Removal using NLTK
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
n=input("Enter a Para:")
normalized_text=n.lower()
tokens=normalized_text.split()
stop_word=set(stopwords.words('english'))
filter=[word for word in tokens if word.isalnum() and word not in stop_word]
print("\nNormalised Text:")
print(normalized_text)
print("\nTokens:")
print(tokens)
print("\nRemove stop words:")
print(filter)
```
